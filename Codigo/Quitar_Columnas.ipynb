```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
Script para copiar archivos CSV eliminando columnas no deseadas.
No recorta por fechas: conserva todas las filas originales.
"""

import os
import pandas as pd

# ============================================================================
# CONFIGURACIÓN
# ============================================================================

# Ruta original donde están los CSVs
ORIGINAL_DIR = os.path.expanduser("/Volumes/copia seguridad1/TFG_Prueba/Datos_TFG_outliers/")

# Ruta donde se guardarán los archivos modificados
OUTPUT_DIR = os.path.expanduser("/Volumes/copia seguridad1/TFG_Prueba/Datos_iniciales/")
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Lista de archivos
station_paths = [
    "T1_E1_Alicante_outliers.csv",
    "T1_E2_Elda_outliers.csv",
    "T2_E1_Elche_outliers.csv",
    "T2_E2_Elda_outliers.csv",
    "T3_E1_Valencia_outliers.csv",
    "T3_E2_Bunol_outliers.csv",
    "T4_E1_Valencia_outliers.csv",
    "T4_E2_Villar_Arzobispo_outliers.csv",
    "T6_E1_Castellon_outliers.csv",
    "T6_E2_Onda_outliers.csv",
    "T7_E1_Sant_Jordi_outliers.csv",
    "T7_E2_Coratxa_outliers.csv",
    "T7_E3_Zorita_outliers.csv",
    "T8_E1_Sant_Jordi_outliers.csv",
    "T8_E2_Morella_outliers.csv",
    "T8_E3_Zorita_outliers.csv",
]


# Columnas a eliminar si existen
columnas_a_eliminar = [
    "Pres.",
    "physical_error",
    "physical_reason",
    "outlier_stl",
    "outlier_jump",
    "outlier_iso",
    "sensor_error",
    "evento_real",
    "doubtful",
]

# ============================================================================
# PROCESAMIENTO
# ============================================================================

def procesar_archivo(nombre_archivo):
    """Carga un CSV, elimina columnas y guarda la versión resultante."""
    ruta_original = os.path.join(ORIGINAL_DIR, nombre_archivo)
    ruta_destino = os.path.join(OUTPUT_DIR, nombre_archivo)

    print(f"Procesando {nombre_archivo}...")

    try:
        # Cargar el CSV
        df = pd.read_csv(ruta_original)

        # Eliminar columnas si existen
        cols_existentes = [col for col in columnas_a_eliminar if col in df.columns]
        if cols_existentes:
            df = df.drop(columns=cols_existentes)
            print(f"  ℹ️ Columnas eliminadas: {cols_existentes}")
        else:
            print("  ℹ️ No se encontraron columnas para eliminar.")

        # Guardar el archivo resultante
        df.to_csv(ruta_destino, index=False)
        print(f"  ✅ Archivo guardado con {len(df)} filas.")

    except Exception as e:
        print(f"  ❌ Error en {nombre_archivo}: {e}")

# Recorrer todos los archivos
for archivo in station_paths:
    procesar_archivo(archivo)

print("\nProceso completado. Revisa la carpeta:")
print(OUTPUT_DIR)
```
